# Get Andrés

```python
import numpy as np
import matplotlib.pyplot as plt

%run "Refraction_index_from_electron_density (1).py"

# Now we just ask for an angle and calculate all the trajectories

print("What frequency do you want?")
freq=float(input())
print("What angle do you want?")
angle=float(input())

n_ld=ref_index(freq,elec_dens_linear_day)
ray=GetRay((angle*np.pi/180),n_ld)

#The values of Ray:
x_axis=[]
y_axis=[]
for j in range(len(ray)):
    x_axis.append(ray[j][0])
    y_axis.append(ray[j][1])
# ONCE WE HAVE THE CORRECT RAY WE DO AN ANDRES
destiny=ray[-1][0]*2
andres=[]
for j in range (len(ray)):
    andres.append([destiny-ray[j][0],ray[j][1]])
x_axis2=[]
y_axis2=[]
for j in range(len(andres)):
    x_axis2.append(andres[j][0])
    y_axis2.append(andres[j][1])

plt.plot(x_axis,y_axis,x_axis2,y_axis2)
```
